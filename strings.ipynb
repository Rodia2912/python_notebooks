String Fundamentals in Python
=============================
> Code snippets from the book :   _Learning Python, Ed 5, by Mark Lutz._ 

### Single- and Double-Quoted Strings Are the Same

```python
'shruberry', 'shruberry'
```

### Implicit concatenation

```python
title = "Meaning " 'of' " Life"
title
```

### Adding commas between these strings would result in a tuple, not a string.

```python
'knight\'s', "knight\"s"
```

### Sequence Operations 

```python
S = 'Spam'
print(len(S))
print(S[0])
print(S[1])
```

```python
print(S)
print(S[1:3])
```

##### Concatenation

```python
S + 'xyz'
```

```python
'xyz' + (S * 9)
```

```python
(S + 'xyz') * 9
```

### Immutability

```python
S[0] = 'z'
# Immutable objects cannot be changed
```

```python
S = 'z' + S[1:]
S  # But we can run expressions to make new objects
```
