# Python: Strings (`str`)
**Guía directa y práctica · Python 3.14**

Un `str` representa texto Unicode. En Python, los strings son **inmutables**: los métodos devuelven un string nuevo, no modifican el original.

Ejecuta las celdas en orden y cambia los ejemplos para practicar.

## 1. Crear strings

Puedes usar comillas simples o dobles. Las triples permiten varias líneas.

```python
nombre = 'Axel'
mensaje = "Hola, mundo"
poema = '''Primera línea
Segunda línea'''

print(nombre)
print(mensaje)
print(poema)
```

Las comillas no forman parte del valor. Para incluir una comilla, puedes alternar tipos o escapar con `\\`.

```python
print("Me llamo 'Axel'")
print('Ella dijo: "hola"')
print("Línea 1\nLínea 2")  # \n = salto de línea
print("Columna 1\tColumna 2")  # \t = tabulador
print("Una barra invertida: \\")
```

### Raw strings y Unicode

- `r"..."` evita que la mayoría de las barras invertidas se interpreten como escapes; útil para rutas y expresiones regulares.
- Los strings admiten Unicode: tildes, ñ, símbolos y emojis.

```python
ruta = r"C:\usuarios\axel\documentos"
print(ruta)
print("Español: ñ, á, ¿qué tal? 🌙")
```

## 2. Unir, repetir y convertir

- `+`: concatenar strings.
- `*`: repetir un string.
- `str(valor)`: convertir un valor a texto.
- `join()`: unir muchos strings con un separador; suele ser la opción adecuada para listas.

```python
print("Hola, " + "Axel")
print("ja" * 3)

edad = 24
print("Edad: " + str(edad))

palabras = ["Python", "es", "útil"]
print(" ".join(palabras))
print(" | ".join(["rojo", "verde", "azul"]))
```

**Importante:** no puedes concatenar directamente `str` e `int` con `+`; convierte el número o usa una f-string.

```python
nombre = "Axel"
edad = 24
print(f"{nombre} tiene {edad} años")
```

## 3. Índices y slicing

Los índices empiezan en `0`. Los negativos cuentan desde el final. En un slice, el final **no se incluye**: `texto[inicio:fin:paso]`.

```python
texto = "Python"

print(texto[0])     # P
print(texto[1])     # y
print(texto[-1])    # n
print(texto[0:2])   # Py
print(texto[2:])    # thon
print(texto[:4])    # Pyth
print(texto[::2])   # Pto
print(texto[::-1])  # nohtyP (invertido)
```

Un índice fuera de rango da `IndexError`; los slices toleran límites fuera de rango. No puedes cambiar un carácter por índice porque los strings son inmutables.

```python
texto = "gato"
# texto[0] = "p"  # TypeError
texto = "p" + texto[1:]
print(texto)  # pato
```

## 4. Longitud, pertenencia y recorrido

- `len(texto)`: número de elementos Unicode del string (no necesariamente el número de bytes ni de caracteres visuales).
- `in` / `not in`: comprobar si aparece una subcadena.
- `for`: recorrer caracteres.

```python
texto = "programar"

print(len(texto))
print("gram" in texto)
print("xyz" not in texto)

for letra in "sol":
    print(letra)
```

## 5. Métodos esenciales

Los métodos no modifican el string original: devuelven otro string.

```python
texto = "  Hola, Mundo!  "

print(texto.lower())       # minúsculas
print(texto.upper())       # MAYÚSCULAS
print(texto.capitalize())  # Primera letra mayúscula, resto minúsculas
print(texto.title())       # Cada palabra con inicial mayúscula
print(texto.swapcase())    # Cambia mayúsculas/minúsculas
print(texto.strip())       # Quita espacios de los extremos
print(texto.lstrip())      # Quita espacios de la izquierda
print(texto.rstrip())      # Quita espacios de la derecha
```

### Buscar y contar

- `find(sub)`: índice de la primera coincidencia o `-1`.
- `index(sub)`: parecido a `find`, pero da `ValueError` si no aparece.
- `count(sub)`: cuenta coincidencias **no solapadas**.
- `startswith()` / `endswith()`: comprueban el principio o el final.

```python
texto = "banana"

print(texto.find("na"))        # 2
print(texto.find("xyz"))       # -1
print(texto.count("an"))       # 2
print(texto.startswith("ba"))  # True
print(texto.endswith("na"))    # True
# texto.index("xyz")  # ValueError
```

### Reemplazar y dividir

- `replace(viejo, nuevo[, cantidad])`: reemplaza coincidencias.
- `split(separador)`: divide y devuelve una lista.
- `split()` sin separador agrupa espacios consecutivos y elimina los extremos.
- `rsplit()` divide empezando por la derecha.
- `partition(separador)`: devuelve una tupla de tres partes: antes, separador y después.

```python
texto = "  Python   es genial  "

print(texto.replace("genial", "útil"))
print(texto.replace("a", "@", 1))
print(texto.split())  # separa por espacios en blanco
print("uno,dos,tres".split(","))
print("a/b/c".rsplit("/", 1))
print("usuario@web.com".partition("@"))
```

### Comprobar el tipo de contenido

Estos métodos devuelven `True` o `False`. Algunos requieren que el string no esté vacío.

```python
print("abc".isalpha())       # solo letras
print("123".isdigit())      # dígitos
print("abc123".isalnum())   # letras o números
print("   ".isspace())      # espacios en blanco
print("hola".islower())     # minúsculas
print("HOLA".isupper())     # mayúsculas
print("Hola Mundo".istitle()) # formato título
print("".isalpha())         # False
```

**Matiz:** `isdigit()` acepta ciertos dígitos Unicode, no solo `0`–`9`. Para validar dígitos ASCII estrictamente, comprueba cada carácter o usa una expresión adecuada.

## 6. Formatear strings

### F-strings (opción habitual)

Pon una `f` antes de las comillas e inserta expresiones entre `{}`. Disponibles desde Python 3.6; Python 3.14 incluye mejoras recientes en su sintaxis.

```python
nombre = "Axel"
precio = 12.5

print(f"Hola, {nombre}")
print(f"En una semana hay {7 * 24} horas")
print(f"Precio: {precio:.2f} €")  # dos decimales
print(f"Porcentaje: {0.756:.1%}") # 75.6%
print(f"{'Python':>10}")          # alineado a la derecha
print(f"{'Python':<10}|")         # alineado a la izquierda
print(f"{42:04d}")                # 0042
```

### Otras opciones

- `format()`: alternativa de formato.
- `%`: formato antiguo; normalmente prefiere f-strings.

```python
print("Hola, {}".format("mundo"))
print("{} + {} = {}".format(2, 3, 5))
print("Nombre: %s, edad: %d" % ("Ana", 20))
```

## 7. Mayúsculas y minúsculas: camelCase, etc.

`lower()` y `upper()` convierten todo el texto. `capitalize()` convierte la primera letra a mayúscula y el resto a minúsculas; `title()` pone iniciales en mayúscula, pero no siempre resuelve bien nombres complejos.

Ejemplo sencillo de `snake_case` o `kebab-case` a `camelCase`:

```python
def to_camel_case(texto):
    palabras = texto.replace("-", " ").replace("_", " ").split()
    if not palabras:
        return ""
    return palabras[0].lower() + "".join(
        palabra.capitalize() for palabra in palabras[1:]
    )

print(to_camel_case("hola mundo"))   # holaMundo
print(to_camel_case("hola_mundo"))   # holaMundo
print(to_camel_case("hola-mundo"))   # holaMundo
```

Este ejemplo es intencionalmente básico: separa por espacios, guiones y guiones bajos, y normaliza las mayúsculas. Para identificadores con acrónimos o reglas especiales puede hacer falta una lógica más específica.

## 8. Prefijos, sufijos y eliminación

`removeprefix()` y `removesuffix()` eliminan un prefijo o sufijo **exacto** si está presente. Están disponibles desde Python 3.9.

```python
archivo = "informe_final.txt"

print(archivo.removesuffix(".txt"))  # informe_final
print(archivo.removeprefix("informe_"))  # final.txt
print("ha-ha-ha".removeprefix("ha-"))  # ha-ha
```

## 9. Escapes y caracteres especiales

| Secuencia | Significado |
|---|---|
| `\n` | salto de línea |
| `\t` | tabulador |
| `\\` | barra invertida |
| `\'` | comilla simple |
| `\"` | comilla doble |
| `\uXXXX` | código Unicode de 4 dígitos hexadecimales |
| `\UXXXXXXXX` | código Unicode de 8 dígitos hexadecimales |

Una barra invertida al final de una línea física puede unir líneas de código; para texto con saltos visibles, suele ser más claro usar `\n` o strings triples.

```python
print("A\nB")
print("\u03A9")  # Ω
print("\U0001F680")  # 🚀
```

## 10. Unicode: detalles útiles

Los strings de Python representan texto Unicode. Dos secuencias distintas pueden verse igual (por ejemplo, una letra acentuada precompuesta o una letra más un acento combinante). Para comparar o normalizar estas formas existe el módulo `unicodedata`.

Además, `len("🚀")` suele ser `1`, pero algunos emojis compuestos visualmente pueden contener varios puntos de código.

```python
import unicodedata

a = "é"
b = "e\u0301"

print(a == b)  # False: representaciones distintas
print(unicodedata.normalize("NFC", a) == unicodedata.normalize("NFC", b))  # True
print(len("🚀"))
print(len("👨‍👩‍👧‍👦"))  # varios puntos de código
```

## 11. Bytes y strings no son lo mismo

`str` es texto; `bytes` son datos binarios. Para convertir entre ellos, especifica una codificación, normalmente UTF-8.

```python
texto = "canción"
datos = texto.encode("utf-8")       # str -> bytes
texto_de_vuelta = datos.decode("utf-8")  # bytes -> str

print(datos)
print(texto_de_vuelta)
```

## 12. Errores típicos

1. **Intentar modificar un carácter:** los strings son inmutables.
2. **Concatenar texto y números:** usa `str()` o una f-string.
3. **Confundir `find()` con `index()`:** `find()` devuelve `-1`; `index()` lanza un error.
4. **Olvidar que el slice excluye el final:** `texto[0:3]` toma índices `0`, `1` y `2`.
5. **Esperar que un método cambie la variable:** guarda el resultado, por ejemplo `texto = texto.strip()`.
6. **Usar `split("")`:** da `ValueError`; para recorrer caracteres usa `list(texto)` o un bucle.
7. **Confundir `strip("abc")` con quitar la subcadena `"abc"`:** `strip()` elimina cualquier combinación de esos caracteres en los extremos. Usa `removeprefix()` / `removesuffix()` para prefijos o sufijos exactos.

```python
texto = "  hola  "
texto.strip()          # devuelve "hola", pero no cambia texto
texto = texto.strip()  # ahora guardas el resultado

print("abcde".strip("ae"))  # "bcd", elimina caracteres de los extremos
print("prefijo_texto".removeprefix("prefijo_"))  # "texto" 
```
