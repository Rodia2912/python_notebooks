### Objetos mutables

#### Listas

```python
L = [2, 3, 4]	# Creamos una lista
M = L           #Hacemos referencia a esa lista 
L[0] = 24       #Hacemos un cambio en la primera
print("L =", L)
print("M =", M)
print("Ambas muestran los mismos datos")
```

En este caso "M" esta haciendo referencia a la lista que guarda "L" por lo que va a mantener cualquier valor que le otorguemos a "L".

```python
L = [2, 3, 4]       #Creamos una lista
M = L[:]		    #Hacemos una copia de L (o list(L), copy.copy(L), etc.)
L[0] = 24           #Cambiamos una valor de su lista

print("L = ", L)    # [24, 3, 4]

print("M = ", M)  
```

### Referencias compartidas e Igualdad

* Operador **==**    igualdad de valores.
* Operador **"is"**  identidad de objetos.

```python
L = [1, 2, 3]
M = L 			    # M y L hacen referencia al mismo Objeto

L == M 			    # True: Same values => operador igualdad de valores
L is M 			    # True: operador identidad de objetos: compara las referencias (los punteros)
```

```python
L = [1, 2, 3]
M = [1, 2, 3]	    # M and L reference different objects

print(
    "L == M",
    L == M 			# True: Mismos valores
)

print(
    "L is M", 	    # False: Objetos diferentes. Operador identidad de objetos: compara las referencias (los punteros)
    L is M
)
```

### Cache
Because small integers and strings are cached and reused, though, is tells us they reference the same single object.

```python
X = 42
Y = 42		    # Should be two different objects

X  == Y         # True
X is Y 		    # True: Same object anyhow: caching at work!!
```
